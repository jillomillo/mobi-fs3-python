```python
import numpy
import matplotlib.pyplot as plt

img = plt.imread('data/cells.png')
T1 = 0.31
T2 = 0.59

img_clip = img.copy()

for row in range(img.shape[0]):
    for col in range(img.shape[1]):
        img[row, col] = img[row, col]

        if img[row, col] < T1:
            img_clip[row, col] = T1
        elif img[row, col] > T2:
            img_clip[row, col] = T2
        else:
            img_clip[row, col] = img[row, col]

plt.imshow(img_clip, 'gray')
plt.colorbar()
```

```python
def clip_image(img_in, T1, T2):
    img_out = img_in.copy()
    img_out[img_out < T1] = T1
    img_out[img_out > T2] = T2
    
    return img_out
```
